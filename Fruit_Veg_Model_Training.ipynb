```python
import os
import zipfile

dataset_zip = zipfile.ZipFile("Fruits_Vegetables.zip")
dataset_root = os.path.join(os.path.expanduser("~"), "fruit_veg_dataset")
```

```python
dataset_zip.extractall(dataset_root)
```

```python
from PIL import Image

Image.open(os.path.join(dataset_root, "Fruits_Vegetables", "train", "cabbage", "Image_11.jpg"))
```

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.models import Sequential
```

```python
pic_height = 180
pic_width = 180
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(dataset_root, "Fruits_Vegetables", "train"),
    image_size=(pic_height, pic_width),
    batch_size=32,
    shuffle=True,
    seed=42,
    validation_split=False
)
```

```python
class_names = train_ds.class_names
```

```python
test_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(dataset_root, "Fruits_Vegetables", "test"),
    image_size=(pic_height, pic_width),
    batch_size=32,
    shuffle=True,
    seed=42,
    validation_split=False
)
```

```python
val_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(dataset_root, "Fruits_Vegetables", "validation"),
    image_size=(pic_height, pic_width),
    batch_size=32,
    shuffle=True,
    seed=42,
    validation_split=False
)
```

```python
cnn_model = Sequential([
    layers.Rescaling(1./255),
    layers.Conv2D(16, 3, padding="same", activation="relu"),
    layers.MaxPooling2D(),
    layers.Conv2D(32, 3, padding="same", activation="relu"),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, padding="same", activation="relu"),
    layers.MaxPooling2D(),
    layers.Dropout(0.2),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(len(class_names))
])
```

```python
cnn_model.compile(optimizer="adam", loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True), metrics=["accuracy"])
```

```python
training_history = cnn_model.fit(train_ds, validation_data=val_ds, epochs=3, batch_size=32)
```

```python
def predict_image(file_path):
    img = tf.keras.utils.load_img(file_path, target_size=(pic_height, pic_width))
    img_array = tf.keras.utils.img_to_array(img)
    img_batch = tf.expand_dims(img_array, 0)
    prediction = cnn_model.predict(img_batch)
    probs = tf.nn.softmax(prediction)
    return f"the provided img is an {class_names[np.argmax(probs)]},and model confidence is {np.max(probs)*100}"
```

```python
predict_image(os.path.join(dataset_root, "Fruits_Vegetables", "test", "lemon", "Image_10.jpg"))
```
